Now we'll look to compare JavaScript and Python conditional statements.

```python
score = 45
```

```python
# A JavaScript conditional statement to compare to a Python version
if(score >= 50) {
    console.log("Pass")
} else {
    console.log("Fail")
}
```

```python
# The equivalent Python code for the above JavaScript conditional statement would be:
if(score >= 50):
    print("Pass")
else:
    print("Fail")
```

Let's test a nested if statement.

```python
# Nested conditional statements in Python
if score >= 70:
    print("Distinction")
elif score >= 50:
    print("Pass")
else:
    print("Resit needed")
```

Here we'll look to create a list and print it to the console. Following this, we'll use the index location of specific values (which starts at 0) and print these.

```python
# Assign values to the variable
skills = ["HTML", "CSS", "JavaScript", "Python", "Machine Learning"]

print(skills) # This will print the entire list of skills
print(skills[0]) # Accessing the first element of the list
print(skills[1]) # Accessing the second element of the list
print(skills[2]) # Accessing the third element of the list
```

Let's test out dictionaries in Python, that let's us store key/value pairs and makes it easy to locate values by calling their keys.

```python
student = {
    "name": "Maya",
    "course": "Python and Data Science",
    "level": 3
}

print(student["name"]) # Accessing the name from the student dictionary
print(student["course"]) # Accessing the course from the student dictionary
print(student["level"]) # Accessing the level from the student dictionary   
```
